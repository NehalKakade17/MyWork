```python
import math
import statistics
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48,1.20, 1.33, 1.18, 1.22, 1.29]
n = len(data)
mean = statistics.mean(data)
s = statistics.stdev(data)  
```

```python
# Part (a): t-interval (sigma unknown)
alpha = 0.01
df = n - 1
```

```python
try:
    from scipy.stats import t
    t_crit = t.ppf(1 - alpha/2, df)
except Exception:
    t_crit = 2.977
se_sample = s / math.sqrt(n)
margin_t = t_crit * se_sample
ci_t_lower = mean - margin_t
ci_t_upper = mean + margin_t
```

```python
# Part (b): z-interval (sigma known = 0.2) ----
sigma = 0.2
# z_{0.995}
try:
    from scipy.stats import norm
    z_crit = norm.ppf(1 - alpha/2)
except Exception:
    z_crit = 2.5758293035489004
se_known = sigma / math.sqrt(n)
margin_z = z_crit * se_known
ci_z_lower = mean - margin_z
ci_z_upper = mean + margin_z
```

```python
print("Sample size:", n)
print(f"Sample mean : {mean}")
print(f"Sample standard deviation : {s}")
print()

print("99% Confidence Interval (using sample s, t-distribution):")
print(f"  t_crit (df={df}): {t_crit}")
print(f"  SE = s/sqrt(n) = {se_sample}")
print(f"  Margin = {margin_t}")
print(f"  CI = ({ci_t_lower}, {ci_t_upper})")
print()

print("99% Confidence Interval (sigma known = 0.2, z-distribution):")
print(f"  z_crit: {z_crit}")
print(f"  SE = sigma/sqrt(n) = {se_known}")
print(f"  Margin = {margin_z}")
print(f"  CI = ({ci_z_lower}, {ci_z_upper})")
print()

print("Interpretation:")
print(" - The t-based 99% CI is wider because we estimated the population SD from the sample.")
```

```python
#Based on the analysis of 15 destructively tested print-heads, the estimated mean durability is about 1.24 million characters before failure.
#Using the sample standard deviation and a t-distribution, the 99% confidence interval for the mean durability is (1.09, 1.39) million characters.
#If the population standard deviation is known to be 0.2, the 99% confidence interval is slightly narrower: (1.11, 1.37) million characters.
#Since both confidence intervals are fairly tight and overlap, we can be 99% confident that the true mean durability lies within this range. 
#The t-interval is wider because it incorporates the additional uncertainty of estimating the population standard deviation from the sample.
#Therefore, the manufacturer can conclude with strong confidence that the average print-head durability is between roughly 1.1 and 1.4 million characters.
```
