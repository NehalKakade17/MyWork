```python
import math 
from scipy.stats import norm
```

```python
# Step 1: State the hypotheses
# H0: μW = μ0 = 4000
# H1: μW > μ0  (costs have increased)
```

```python
sample_mean = 3050 # observed mean
X_mean = 600       # average
mu0 = 1000 + 5 * X_mean   # theoretical mean
sigma_X = 25       # std dev of X
sigma_W = 5 * sigma_X     # std dev of W
n = 25             # sample size
```

```python
# Standard error
se = sigma_W / math.sqrt(n)
```

```python
# Test statistic (z)
z = (sample_mean - mu0) / se
```

```python
#Step 3: Critical value and p-value
alpha = 0.05
z_critical = norm.ppf(1 - alpha)   
p_value = 1 - norm.cdf(z)         
```

```python
# Step 4: Decision
if z > z_critical:
    decision = "Reject H0: Evidence suggests costs have increased."
else:
    decision = "Fail to reject H0: No evidence that costs have increased."
```

```python
# Output Results
print("Hypothesis Testing Results")
print(f"Theoretical mean: {mu0}")
print(f"Sample mean: {sample_mean}")
print(f"Standard deviation of W: {sigma_W}")
print(f"Sample size (n): {n}")
print(f"Standard error (SE): {se:.2f}")
print(f"Test Statistic (z): {z:.2f}")
print(f"Critical Value: {z_critical:.2f}")
print(f"P-value: {p_value:.4f}")
print("Conclusion:", decision)
```

```python
There is no strong evidence to support the restaurant owners’ claim that the weekly operating costs are higher than the theoretical model
(W=1000+5X) predicts.
In fact, the observed sample mean (3050) is much lower than the model’s predicted cost (4000), which contradicts the claim of increased costs.
```
